# Insurance Charges Prediction Using Machine Learning
This notebook loads the insurance dataset, preprocesses the data, trains multiple regression models, and compares their R² scores.

```python
# Import Required Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score
```

```python
#Load the Dataset 
df = pd.read_csv("insurance_pre.csv")
```

```python
# Printing the dateset
df
```

```python
# Display the Dataset
df.head()
```

```python
# Check Dataset Information
df.info()
```

```python
# Check Dataset Shape
print("Number of Rows :", df.shape[0])
print("Number of Columns :", df.shape[1])
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Statistical Summary
df.describe()
```

```python
# Separate Features and Target Variable
X = df.drop("charges", axis=1)
y = df["charges"]
```

```python
y
```

```python
# checking the first five values from dataset apart from charges
X.head()
```

```python
# Display Target Variable
y.head()
```

```python
# Identify Categorical and Numerical Columns
categorical_columns = ["sex", "smoker"]
numerical_columns = ["age", "bmi", "children"]
```

```python

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(drop="first"), categorical_columns)
    ],
    remainder="passthrough"
)
```

```python
# Perform One-Hot Encoding
df=pd.get_dummies(df,drop_first=True)
```

```python
df
```

```python
# Split the Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=0
)
```

```python
# Create Multiple Linear Regression Model
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
# Train the Model
model.fit(X_train, y_train)
```

```python
# Display the Bias (Intercept)
bias = model.named_steps["regressor"].intercept_

print("Bias (Intercept):")
print(bias)
```

```python
# Display the Weights
weights = model.named_steps["regressor"].coef_

print("Weights (Coefficients):")
print(weights)
```

```python
# Display Feature Names with Their Weights
feature_names = model.named_steps["preprocessor"].get_feature_names_out()

coefficients = pd.DataFrame({
    "Feature": feature_names,
    "Weight": weights
})

coefficients
```

```python
# Predict the Test Data
y_pred = model.predict(X_test)
```

```python
# Calculate the R² Score
r2 = r2_score(y_test, y_pred)
```

```python
r2
```

```python
import pickle 
filename="finalized_model_linear_regression.sav"
with open(filename, "wb") as file:
    pickle.dump(model, file)

print("Model saved successfully!")
```

```python
filename = "finalized_model_linear_regression.sav"

with open(filename, "rb") as file:
    loaded_model = pickle.load(file)
    new_customer = pd.DataFrame({
    "age": [35],
    "sex": ["male"],
    "bmi": [27.5],
    "children": [2],
    "smoker": ["no"]
})
prediction = loaded_model.predict(new_customer)
print("Predicted Insurance Charges:", prediction[0])
```
